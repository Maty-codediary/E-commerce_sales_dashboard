```python
import pandas as pd
```

```python
sales = pd.read_csv('data/processed/sales.csv')
sales.dtypes
```

```python
orders_status = pd.read_csv('data/processed/orders_status.csv')
orders_status.dtypes
```

```python
sales['purchase_at'] = pd.to_datetime(sales['purchase_at'])
```

```python
orders_status['purchase_at'] = pd.to_datetime(orders_status['purchase_at'])
```

```python
delivered = sales['order_status'] == 'delivered'
```

```python
total_revenue = sales[delivered]['revenue'].sum()
```

```python
orders = sales[delivered]['order_id'].nunique()
orders
```

```python
AOV = total_revenue/orders
AOV
```

```python
total_customers = sales[delivered]['customer_unique_id'].nunique()
total_customers
```

```python
monthly_sales = sales[delivered].groupby('order_month')['revenue'].sum().loc['2017-01':]
```

```python
growth = (monthly_sales.pct_change() * 100).round(1)
growth.head()
```

```python
growth.tail()
```

```python
sales.columns
```

```python
top_10_by_categories = sales[delivered].groupby('category')['revenue'].sum().sort_values(ascending=False).head(10)
top_10_by_categories
```

```python
top_10_by_state = sales[delivered].groupby('customer_state')['revenue'].sum().sort_values(ascending=False).head(10)
top_10_by_state
```

```python
top_state_share = (top_10_by_state/total_revenue)*100
top_state_share
```

```python
orders_by_status = orders_status['order_status'].value_counts(normalize=True) * 100
orders_by_status
```

```python
orders_by_state = sales[delivered].groupby('customer_state').agg(
    median_days=('delivery_days', 'median'),
    late_rate = ('is_late', 'mean'),
    orders=('order_id', 'nunique')
).sort_values(by= "median_days", ascending=False)
```

```python
orders_by_state['late_rate'] = orders_by_state['late_rate'] * 100
```
